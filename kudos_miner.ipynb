이 노트북을 실행하면 구글 코랩의 고성능 GPU를 이용해 **AI Horde 분산 네트워크 워커**를 가동합니다.
다른 사람의 AI 생성 요청을 초고속으로 처리해주면서 **수만~수십만 단위의 Kudos(우선순위 포인트)를 대량 채굴**합니다.

채굴된 Kudos는 본인 계정에 즉시 적립되어, 내 TRPG 게임에서 이미지 생성 시 **대기열 0초(VIP 최우선 순위)**로 즉시 생성됩니다!

### 📌 사용 방법
1. 상단 메뉴에서 **[런타임] → [런타임 유형 변경] → 하드웨어 가속기: [T4 GPU / A100 GPU]** 확인
2. 아래 **[1단계]**에 본인의 **AI Horde API Key**를 입력
3. 상단 메뉴에서 **[런타임] → [모두 실행]** (또는 각 셀의 재생 버튼 순서대로 클릭)

```python
# [1단계] GPU 사양 확인
!nvidia-smi
```

```python
# [2단계] 최신 KoboldCpp 고성능 CUDA 바이너리 다운로드 (약 10초 소요)
import os

!wget -q -O koboldcpp https://github.com/LostRuins/koboldcpp/releases/latest/download/koboldcpp-linux-x64
!chmod +x koboldcpp
print("✅ KoboldCpp 바이너리 준비 완료!")
```

```python
# [3단계] 전 세계에서 요청량이 가장 많은 인기 Llama-3-8B 모델 다운로드
# (요청량이 많을수록 쉬지 않고 처리하여 Kudos가 폭풍 적립됩니다)
import os

MODEL_FILE = "Meta-Llama-3-8B-Instruct.Q4_K_M.gguf"
if not os.path.exists(MODEL_FILE):
    print("📥 모델 다운로드 중 (약 30초~1분 소요)... 4.9GB")
    !wget -q -O {MODEL_FILE} https://huggingface.co/QuantFactory/Meta-Llama-3-8B-Instruct-GGUF/resolve/main/Meta-Llama-3-8B-Instruct.Q4_K_M.gguf
    print("✅ 모델 다운로드 완료!")
else:
    print("✅ 기존 모델 파일 발견!")
```

```python
# [4단계] AI Horde 채굴 시작!
# ⚠️ 아래 API_KEY에 https://aihorde.net/register 에서 발급받은 본인 키를 입력하세요.

API_KEY = "여기에_본인의_AI_HORDE_API_KEY_입력"  # 예: "abcdef123456..."
WORKER_NAME = "MyColabProWorker"                   # 원하는 워커 이름 (영문/숫자, 아무거나)

if API_KEY == "여기에_본인의_AI_HORDE_API_KEY_입력" or not API_KEY:
    raise ValueError("🚨 [오류] API_KEY에 발급받으신 본인의 AI Horde API 키를 입력해 주세요!")

print(f"🔥 AI Horde 채굴을 시작합니다! (워커: {WORKER_NAME})")
print("💡 요청이 들어올 때마다 자동으로 처리되며 Kudos가 실시간으로 쌓입니다!")

# 백그라운드 끊김 방지 오디오 실행 + 코볼드CPP 호드 워커 기동
!./koboldcpp --model {MODEL_FILE} --usecublas normal --gpulayers 99 --contextsize 4096 --hordekey {API_KEY} --hordeworkername {WORKER_NAME} --hordemodelname "Llama-3-8B-Instruct"
```
