```python
# [3단계] 전 세계에서 요청량이 가장 많은 인기 Llama-3-8B 모델 다운로드
import os

MODEL_FILE = "Meta-Llama-3-8B-Instruct.Q4_K_M.gguf"
if not os.path.exists(MODEL_FILE):
    print("📥 모델 다운로드 중 (약 30초~1분 소요)... 4.9GB")
    !wget -q -O {MODEL_FILE} https://huggingface.co/QuantFactory/Meta-Llama-3-8B-Instruct-GGUF/resolve/main/Meta-Llama-3-8B-Instruct.Q4_K_M.gguf
    print("✅ 모델 다운로드 완료!")
else:
    print("✅ 기존 모델 파일 발견!")
```

```python
# [4단계] AI Horde 텍스트 채굴 시작!
# ⚠️ 상대방의 요청과 내 AI의 답변이 화면 출력 및 horde_dialogues.log 파일에 자동 저장됩니다.

API_KEY = "L4qKfKC37bLUDD5hgOSeSg"
WORKER_NAME = "YeolPhaCham_Scribe_01"

print(f"🔥 AI Horde 텍스트 채굴을 시작합니다! (워커: {WORKER_NAME})")
print("💡 요청이 들어올 때마다 대답 내용이 'horde_dialogues.log'에 자동 기록됩니다!")

# 코볼드CPP 호드 워커 기동 + 실시간 화면 출력 및 로그 파일(horde_dialogues.log) 누적 저장
!./koboldcpp --model {MODEL_FILE} --usecublas normal --gpulayers 99 --contextsize 4096 --hordekey {API_KEY} --hordeworkername {WORKER_NAME} --hordemodelname "Llama-3-8B-Instruct" 2>&1 | tee -a horde_dialogues.log
```
